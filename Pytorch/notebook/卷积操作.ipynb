# 卷积操作

**torch.nn**

**卷积层原理conv**

```python
import torch
import torch.nn as nn

conv = nn.Conv2d(
    in_channels=3,
    out_channels=6,
    kernel_size=3,
    stride=1,
    padding=0
)

'''
in_channels: 输入图片的通道数。RGB 图片是 3，灰度图片通常是 1
out_channels: 卷积核的数量，也是输出特征图的通道数
kernel_size: 卷积核大小，3 表示 3×3
stride: 卷积核每次移动的距离，横向和纵向都移动 1 个像素
padding: 输入四周补零的层数；0 表示不补零

'''
```

搭建普通神经网络时使用 `nn.Conv2d`，因为它会自动创建、保存并训练卷积核和偏置，可以方便地写在模型的 `__init__` 中；只有当你需要手动提供或动态生成卷积核、共享已有权重，或者实现特殊卷积计算时，才使用 `torch.nn.functional.conv2d`。简单记忆：**需要一个可训练的卷积层就用 `nn.Conv2d`，已经有具体卷积核并只想执行卷积运算就用 `F.conv2d`。**
